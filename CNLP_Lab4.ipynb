4.1. To perform Term-Frequency Analysis and Named Entity Recognition on a given text using any toolkit.
●	Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
●	Identify the top 10 most frequent terms and display their corresponding frequencies.
.


```python
import nltk
import re
import csv
from collections import Counter

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('maxent_ne_chunker')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
nltk.download('averaged_perceptron_tagger_eng')

from nltk.tokenize import word_tokenize
from nltk import pos_tag, ne_chunk

# Read input file
with open("/content/drive/MyDrive/CNLP-Lab/Experiment-4/4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()

text_lower = text.lower()

words = word_tokenize(text_lower)

words = [word for word in words if word.isalpha()]

frequency = Counter(words)

with open("term_frequency_4_1.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(["Term", "Frequency"])

    for word, count in frequency.most_common():
        writer.writerow([word, count])


print("Top 10 Most Frequent Terms:")
for word, count in frequency.most_common(10):
    print(word, ":", count)



tokens = word_tokenize(text)
pos_tags = pos_tag(tokens)

named_entities = ne_chunk(pos_tags)

print("\nNamed Entities:")

for entity in named_entities:
    if hasattr(entity, "label"):
        name = " ".join(word for word, tag in entity)
        print(name, "->", entity.label())
```

4.2. To perform Term-Frequency Analysis on a given text without using any toolkit.

●	Calculate the frequency of each word and save the output in CSV format (Term, Frequency)
●	Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
import re
import csv
from collections import Counter


with open("/content/drive/MyDrive/CNLP-Lab/Experiment-4/4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()

text = text.lower()

text = re.sub(r'[^\w\s]', '', text)

words = text.split()

frequency = Counter(words)

with open("term_frequency_4_2.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(["Term", "Frequency"])

    for word, count in frequency.most_common():
        writer.writerow([word, count])

print("Top 10 Most Frequent Terms:")

for word, count in frequency.most_common(10):
    print(word, ":", count)
```

4.3. Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:
1.	Read multiple text documents.

2.	Convert text to lowercase.

3.	Remove punctuation.

4.	Tokenize the documents.

5.	Calculate TF for every term.

6.	Calculate DF for every term.

7.	Calculate IDF using the logarithmic formula.

8.	Calculate TF-IDF for every term in every document.

9.	Display Term, TF, DF, IDF, and TF-IDF at each step.

10.	Identify the top 10 terms with the highest TF-IDF score for each document.


```python
import re
import math
from collections import Counter


documents = {
    "Document 1": "Natural language processing is a field of artificial intelligence.",

    "Document 2": "Natural language processing helps computers understand human language.",

    "Document 3": "Machine learning is an important part of artificial intelligence."
}


processed_documents = {}

for doc_name, text in documents.items():

    text = text.lower()

    text = re.sub(r'[^\w\s]', '', text)

    words = text.split()

    processed_documents[doc_name] = words



tf = {}

for doc_name, words in processed_documents.items():

    word_count = Counter(words)

    total_words = len(words)

    tf[doc_name] = {}

    for word, count in word_count.items():

        tf[doc_name][word] = count / total_words



df = {}

for doc_name, words in processed_documents.items():

    unique_words = set(words)

    for word in unique_words:

        if word in df:
            df[word] += 1
        else:
            df[word] = 1



number_of_documents = len(documents)

idf = {}

for word, document_frequency in df.items():

    idf[word] = math.log(
        number_of_documents / document_frequency
    )



tf_idf = {}

for doc_name in processed_documents:

    tf_idf[doc_name] = {}

    for word in tf[doc_name]:

        tf_idf[doc_name][word] = (
            tf[doc_name][word] * idf[word]
        )



print("\n========== TERM FREQUENCY ==========\n")

for doc_name in tf:

    print(doc_name)

    for word, value in tf[doc_name].items():

        print(
            word,
            "TF =",
            round(value, 4)
        )

    print()


print("\n========== DOCUMENT FREQUENCY ==========\n")

for word, value in df.items():

    print(
        word,
        "DF =",
        value
    )


print("\n========== INVERSE DOCUMENT FREQUENCY ==========\n")

for word, value in idf.items():

    print(
        word,
        "IDF =",
        round(value, 4)
    )


print("\n========== TF-IDF ==========\n")

for doc_name in tf_idf:

    print(doc_name)

    for word, value in tf_idf[doc_name].items():

        print(
            word,
            "TF-IDF =",
            round(value, 4)
        )

    print()


print("\n========== TOP 10 TF-IDF TERMS ==========\n")

for doc_name in tf_idf:

    sorted_terms = sorted(
        tf_idf[doc_name].items(),
        key=lambda x: x[1],
        reverse=True
    )

    print(doc_name)

    for word, score in sorted_terms[:10]:

        print(
            word,
            "->",
            round(score, 4)
        )

    print()
```
