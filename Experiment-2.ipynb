# Experiment 2.1:  Write a Python program to perform basic text preprocessing on a given data. The program should:
1.	Convert text to lowercase. (Also display how many uppercase?)
2.	Remove punctuation.        (Also display how many different types of punctuations?)
3.	Remove numbers.            (Also display how many numbers?)
4.	Remove extra whitespace. (How much extra whitespace is present?)
5.	Display the cleaned text.


```python
import re
import string

file_path = '/content/drive/MyDrive/CNLP-Lab/Experiment-2/2.1_text_data.txt'

try:
    with open(file_path, 'r', encoding='utf-8') as f:
        original_text = f.read()
    print("Original text loaded successfully. First 200 characters:")
    print(original_text[:200])
except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found. Please ensure the path is correct.")
    original_text = ""

processed_text = original_text
```

```python
uppercase_chars_count = sum(1 for char in processed_text if char.isupper())
print(f"Number of uppercase characters before conversion: {uppercase_chars_count}")

processed_text = processed_text.lower()
print("Text converted to lowercase. First 200 characters:")
print(processed_text[:200])
```

```python
punctuation_counts = {}
for char in processed_text:
    if char in string.punctuation:
        punctuation_counts[char] = punctuation_counts.get(char, 0) + 1

print("Counts of different punctuation types:")
if punctuation_counts:
    for punc, count in punctuation_counts.items():
        print(f"  '{punc}': {count}")
else:
    print("  No punctuation found.")

processed_text = ''.join([char for char in processed_text if char not in string.punctuation])
print("Punctuation removed. First 200 characters:")
print(processed_text[:200])
```

```python
number_matches = re.findall(r'\d+', processed_text)
total_numbers_found = len(number_matches)
print(f"Total individual numbers found before removal: {total_numbers_found}")

processed_text = re.sub(r'\d+', '', processed_text)
print("Numbers removed. First 200 characters:")
print(processed_text[:200])
```

```python
extra_whitespace_blocks = re.findall(r'\s{2,}', processed_text)
num_extra_whitespace_blocks = len(extra_whitespace_blocks)
print(f"Number of blocks of extra whitespace (two or more consecutive spaces): {num_extra_whitespace_blocks}")

extra_whitespace_chars = sum(len(match.group(0)) - 1 for match in re.finditer(r'\s+', processed_text) if len(match.group(0)) > 1)
print(f"Total number of individual 'extra' whitespace characters removed: {extra_whitespace_chars}")

processed_text = re.sub(r'\s+', ' ', processed_text).strip()
print("Extra whitespace removed. First 200 characters:")
print(processed_text[:200])
```

```python
print("Final Cleaned Text (first 500 characters):")
print(processed_text[:500])

if len(processed_text) < 2000:
    print("\nFull Cleaned Text:")
    print(processed_text)
else:
    print(f"\nFull cleaned text is too long ({len(processed_text)} characters) to display in full. Displayed first 500 characters.")
```

# Experiment 2.2: Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.
           (Three separate code: using NLTK, using spaCy  and without using NLTK/spacy)


```python
file_path_2_2 = '/content/drive/MyDrive/CNLP-Lab/Experiment-2/2.2_tokenization_data.txt'

try:
    with open(file_path_2_2, 'r', encoding='utf-8') as f:
        text_for_tokenization = f.read()
    print("Text loaded successfully for tokenization. First 200 characters:")
    print(text_for_tokenization[:200])
except FileNotFoundError:
    print(f"Error: The file '{file_path_2_2}' was not found. Please ensure the path is correct.")
    text_for_tokenization = ""
```

## 1. Tokenization using NLTK

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize, sent_tokenize

if text_for_tokenization:
    nltk_sentences = sent_tokenize(text_for_tokenization)
    print("\n--- NLTK Sentence Tokenization ---")
    for i, sentence in enumerate(nltk_sentences):
        print(f"Sentence {i+1}: {sentence}")

    nltk_words = word_tokenize(text_for_tokenization)
    print("\n--- NLTK Word Tokenization ---")
    print(nltk_words[:50])
    print(f"Total words (NLTK): {len(nltk_words)}")
else:
    print("No text available for NLTK tokenization.")
```

## 2. Tokenization using spaCy

```python
!pip install spacy
!python -m spacy download en_core_web_sm

import spacy

try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    print("Downloading spaCy model 'en_core_web_sm'...")
    !python -m spacy download en_core_web_sm
    nlp = spacy.load("en_core_web_sm")

if text_for_tokenization:
    doc = nlp(text_for_tokenization)

    spacy_sentences = [sent.text for sent in doc.sents]
    print("\n--- spaCy Sentence Tokenization ---")
    for i, sentence in enumerate(spacy_sentences):
        print(f"Sentence {i+1}: {sentence}")

    spacy_words = [token.text for token in doc]
    print("\n--- spaCy Word Tokenization ---")
    print(spacy_words[:50])
    print(f"Total words (spaCy): {len(spacy_words)}")
else:
    print("No text available for spaCy tokenization.")
```

## 3. Tokenization without NLTK/spaCy (Manual Approach)

```python
import re

if text_for_tokenization:
    manual_sentences = re.split(r'(?<=[.!?])\s+', text_for_tokenization.strip())
    print("\n--- Manual Sentence Tokenization ---")
    for i, sentence in enumerate(manual_sentences):
        print(f"Sentence {i+1}: {sentence}")

    temp_text = text_for_tokenization.lower()
    temp_text = re.sub(r'[^\w\s]', '', temp_text)
    manual_words = temp_text.split()

    print("\n--- Manual Word Tokenization (cleaned text) ---")
    print(manual_words[:50])
    print(f"Total words (Manual): {len(manual_words)}")
else:
    print("No text available for manual tokenization.")
```

#Experiment 2.3: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).

```python
file_path_2_3 = '/content/drive/MyDrive/CNLP-Lab/Experiment-2/2.2_tokenization_data.txt' # Assuming the same data as 2.2 for tokenization base

try:
    with open(file_path_2_3, 'r', encoding='utf-8') as f:
        text_for_stopword_removal = f.read()
    print("Text loaded successfully for stop word removal. First 200 characters:")
    print(text_for_stopword_removal[:200])
except FileNotFoundError:
    print(f"Error: The file '{file_path_2_3}' was not found. Please ensure the path is correct.")
    text_for_stopword_removal = ""
```

### Stop Word Removal using NLTK

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('stopwords') # Download NLTK's stop words list

if text_for_stopword_removal:
    stop_words = set(stopwords.words('english'))
    original_tokens = word_tokenize(text_for_stopword_removal.lower())
    filtered_tokens = [word for word in original_tokens if word.isalnum() and word not in stop_words]
    removed_stop_words = sorted(list(set(word for word in original_tokens if not word.isalnum() or word in stop_words)))

    print("\n--- Original Tokens (first 50) ---")
    print(original_tokens[:50])
    print(f"Total original tokens: {len(original_tokens)}")

    print("\n--- Stop Words Identified and Removed (unique) ---")
    print(removed_stop_words)
    print(f"Total unique stop words removed: {len(removed_stop_words)}")

    print("\n--- Filtered Tokens (first 50) ---")
    print(filtered_tokens[:50])
    print(f"Total filtered tokens: {len(filtered_tokens)}")
else:
    print("No text available for stop word removal.")
```
