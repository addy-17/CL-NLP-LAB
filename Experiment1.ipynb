## Experiment-1.1: Read the employee dataset and display the average salary of each department using a bar chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/drive/MyDrive/CNLP-Lab/Experiment-1/employee_information_100.csv')
df.head()


average_salary_per_department = df.groupby('Department')['Salary'].mean()
print("Average salary per department:")
display(average_salary_per_department)
```

```python
fig = plt.figure(figsize=(10, 6))
average_salary_per_department_df = average_salary_per_department.reset_index()
sns.barplot(x='Department', y='Salary', data=average_salary_per_department_df, palette='viridis', hue='Department', legend=False)
plt.title('Average Salary per Department')
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

## Experiment 1.2: Display the number of employees working in each department.

```python
employees_per_department = df.groupby('Department')['Employee_ID'].count()
print("Number of employees per department:")
display(employees_per_department)
```

```python
fig = plt.figure(figsize=(10, 6))
employees_per_department_df = employees_per_department.reset_index()
sns.barplot(x='Department', y='Employee_ID', data=employees_per_department_df, palette='viridis', hue='Department', legend=False)
plt.title('Number of Employees per Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

## Experiment 1.3:Display the percentage of male and female employees using a pie chart


```python
gender_count = df["Gender"].value_counts()

print(gender_count)

plt.figure(figsize=(6,6))
gender_count.plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Gender Distribution")
plt.ylabel("")
plt.show()
```

## Experiment 1.4: Draw a histogram showing salary distribution.

```python
plt.figure(figsize=(8,5))
plt.hist(df["Salary"], bins=10)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

## Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.


```python
plt.figure(figsize=(8,5))
plt.scatter(df["Experience_Years"], df["Salary"])

plt.title("Experience vs Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.show()
```

## Experiment 1.6: Display the top 10 employees with the highest salary.

```python
top10 = df.nlargest(10, "Salary")

print(top10[["Employee_ID", "Name", "Department", "Salary"]])
```

## Experiment 1.7: Find the highest salary in every department.


```python
highest_salary = df.groupby("Department")["Salary"].max()

print(highest_salary)
```

## Experiment 1.8: Display employees whose salary is greater than the overall average salary.


```python
average_salary = df["Salary"].mean()

print("Overall Average Salary =", average_salary)

employees = df[df["Salary"] > average_salary]

print(employees)
```

## Experiment 1.9: Calculate the average years of experience in each department.

```python
avg_exp = df.groupby("Department")["Experience_Years"].mean()

print(avg_exp)
```
