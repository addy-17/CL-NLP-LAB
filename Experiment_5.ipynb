## Experiment 5.1. Implement subword-level tokenization using a BPE-based tokenizer.
- Tokenize the sentence into subword units.
- Display the generated tokens and their corresponding token IDs.

(Implement two separate codes: a. Using pretrained model and b. Without using pretrained model)



```python
from transformers import GPT2Tokenizer

# Load input data
with open("/content/drive/MyDrive/CNLP-Lab/Experiment-5/input_sub_word_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Load pretrained GPT-2 BPE tokenizer
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

# Tokenize the text
tokens = tokenizer.tokenize(text)

# Convert tokens to token IDs
token_ids = tokenizer.convert_tokens_to_ids(tokens)

# Display tokens and IDs
print("Total Tokens:", len(tokens))
print("\nTokens and Token IDs:")
print("-" * 50)

for token, token_id in zip(tokens, token_ids):
    print(f"{token:25} -> {token_id}")


from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

# Read input file
with open("/content/drive/MyDrive/CNLP-Lab/Experiment-5/input_sub_word_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Create BPE tokenizer
tokenizer = Tokenizer(BPE(unk_token="[UNK]"))

# Use whitespace pre-tokenization
tokenizer.pre_tokenizer = Whitespace()

# Define BPE trainer
trainer = BpeTrainer(
    vocab_size=1000,
    min_frequency=2,
    special_tokens=["[UNK]"]
)

# Train tokenizer from the input corpus
tokenizer.train_from_iterator([text], trainer)

# Test sentence
test_text = "Natural language processing is a branch of artificial intelligence."

# Encode the sentence
output = tokenizer.encode(test_text)

# Display tokens and IDs
print("Tokens and Token IDs:")
print("-" * 50)

for token, token_id in zip(output.tokens, output.ids):
    print(f"{token:25} -> {token_id}")

print("\nVocabulary Size:", tokenizer.get_vocab_size())
```

 ## Experiment 5.3. Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, and description of each POS tag.
                            
"The young student is reading an interesting book in the library."   

(Implement two separate codes: a. Using Spacy and b.  NLTK )


```python
import spacy
import nltk

# Load English language model
nlp = spacy.load("en_core_web_sm")

# Input sentence
text = "The young student is reading an interesting book in the library."

# Process text
doc = nlp(text)

print("Token\t\tPOS\t\tDescription")
print("-" * 60)

# Set for unique tokens
seen = set()

for token in doc:
    if token.text not in seen:
        print(
            f"{token.text:15} "
            f"{token.pos_:10} "
            f"{spacy.explain(token.pos_)}"
        )
        seen.add(token.text)


nltk.download('punkt_tab') # Add this line to download the required resource
nltk.download('averaged_perceptron_tagger_eng') # Add this line to download the required resource

text = "The young student is reading an interesting book in the library."

# Tokenize
tokens = nltk.word_tokenize(text)

# POS tagging
pos_tags = nltk.pos_tag(tokens)

# POS descriptions
pos_descriptions = {
    "DT": "Determiner",
    "JJ": "Adjective",
    "JJR": "Comparative adjective",
    "JJS": "Superlative adjective",
    "NN": "Noun",
    "NNS": "Plural noun",
    "VB": "Verb",
    "VBD": "Past tense verb",
    "VBG": "Gerund/Present participle",
    "VBN": "Past participle",
    "VBP": "Present tense verb",
    "VBZ": "Third-person singular verb",
    "IN": "Preposition/Subordinating conjunction",
    "CC": "Coordinating conjunction",
    "PRP": "Personal pronoun",
    "RB": "Adverb",
    ".": "Punctuation"
}

print("Token\t\tPOS\t\tDescription")
print("-" * 60)

seen = set()

for token, tag in pos_tags:
    if token not in seen:
        description = pos_descriptions.get(tag, "Other")

        print(
            f"{token:15} "
            f"{tag:10} "
            f"{description}"
        )

        seen.add(token)
```

## Experiment 5.4. Implement Part-of-Speech (POS) tagging on a given text and display the unique token, POS tag, description of each token and frequency using Spacy.

```python
import spacy
from collections import Counter

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

# Read corpus
with open("/content/drive/MyDrive/CNLP-Lab/Experiment-5/input_sub_word_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

# POS tagging
doc = nlp(text)

# Store unique token information
token_info = {}

for token in doc:

    if token.is_space:
        continue

    if token.text not in token_info:
        token_info[token.text] = {
            "POS": token.pos_,
            "Description": spacy.explain(token.pos_),
            "Frequency": 0
        }

    token_info[token.text]["Frequency"] += 1


# Sort according to frequency
sorted_tokens = sorted(
    token_info.items(),
    key=lambda x: x[1]["Frequency"],
    reverse=True
)

# Display
print("Token\t\tPOS\t\tDescription\t\tFrequency")
print("-" * 90)

for token, info in sorted_tokens:

    print(
        f"{token:18} "
        f"{info['POS']:10} "
        f"{str(info['Description']):30} "
        f"{info['Frequency']}"
    )
```
